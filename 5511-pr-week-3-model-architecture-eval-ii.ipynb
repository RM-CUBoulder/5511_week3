# Evaluation Model Framework 3
#### Code for evaluating three model architectures within Model Framework 3.

```python
# Import necessary modules
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
from os import listdir
import matplotlib.pyplot as plt
```

```python
# Function to plot training metrics against the number of epochs
def pd_plot(metric_df,title, path):

    fig, ax = plt.subplots(3, 1, figsize=(8,6))
    ax[0].plot(metric_df.accuracy, linestyle= 'dashed')
    ax[0].plot(metric_df.val_accuracy, linestyle= 'dotted')
    ax[0].legend(["Accuracy", "Val. Accuracy"], loc="lower right")
    ax[0].set_ylabel('Accuracy')
    ax[0].set_ylim(0,1)
    ax[0].set_xticks([0, 4, 8, 12, 16, 20], ['0', '4', '8', '12', '16', '20'])
    
    ax[1].plot(metric_df.auc, linestyle= 'dashed')
    ax[1].plot(metric_df.val_auc, linestyle= 'dotted')
    ax[1].legend(["AUC", "Val. AUC"], loc="lower right")
    ax[1].set_ylabel('AUC')
    ax[1].set_ylim(0,1)
    ax[1].set_xticks([0, 4, 8, 12, 16, 20], ['0', '4', '8', '12', '16', '20'])
    
    ax[2].plot(metric_df.loss, linestyle= 'dashed')
    ax[2].plot(metric_df.val_loss, linestyle= 'dotted')
    ax[2].legend(["Loss", "Val. Loss"], loc="lower right")
    ax[2].set_ylabel('Loss')
    ax[2].set_xticks([0, 4, 8, 12, 16, 20], ['0', '4', '8', '12', '16', '20'])
    ax[2].set_xlabel('# Epoch', fontsize = 12)


    ax[0].set_title(title, fontsize = 16)
    
    fig.tight_layout()
    #plt.show()
    plt.savefig(path)
```

```python
# Read Architecture 1 Metrics
output = pd.read_csv('Simple_CNN_St_no.csv')
output
```

```python
# Plot Architecture 1 Metrics
pd_plot(output, 'Simple CNN (Stan. Norm.) training metrics', 'simple_cnn_st_no.jpeg')
```

```python
# Read Architecture 2 Metrics
output1 = pd.read_csv('Basic_CNN_St_no.csv')
output1
```

```python
# Plot Architecture 2 Metrics
pd_plot(output1, 'Basic CNN (Stan. Norm.) training metrics', 'basic_cnn_st_no.jpeg')
```

```python
# Read Architecture 3 Metrics
output2 = pd.read_csv('Basic_CNN_St_no_20per.csv')
output2
```

```python
# Plot Architecture 3 Metrics
pd_plot(output2, 'Basic CNN (Stan. Norm. 20%) training metrics','basic_cnn_st_no_20per.jpeg')
```
