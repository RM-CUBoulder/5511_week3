# Model Scoring 
#### Code for generating probabilities for the Histopathologic Cancer Detection data set with previously created Convolutional Neural Networks.

```python
# Import necessary modules
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import matplotlib.pyplot as plt
import os
from os import listdir
from os.path import isfile, join
from PIL import Image
import keras
from keras import layers
import csv
from tensorflow.keras.utils import to_categorical
import tensorflow as tf
```

```python
# Read all .tif files from the 'test' folder
test_obs = []
for dirname, _, filenames in os.walk('/kaggle/input/histopathologic-cancer-detection/test'):
    for filename in filenames:
        test_obs.append(filename)
```

```python
# Function to transform .tif file to an array
def to_array(test_obs = test_obs):

    # String of working directory
    test_dir = '/kaggle/input/histopathologic-cancer-detection/test'
    
    imarray_totaal = []
    list_df = []
    # Loop to import test images in list_df (by file names) from working directory
    for j in test_obs:

        im = Image.open(test_dir + '/' + j )

        # Turn '.tif' file into array
        imarray = np.array(im)
        
        imarray_totaal.append(imarray/255) # Normalize RGB values
        list_df.append(filename)
    
    # Turn list into array
    imarray_totaal= np.array(imarray_totaal)
    
    # Giving imarray_totaal (x_train) and labels (y_train) familiar names 
    x_test = imarray_totaal   

    return x_test
```

```python
# Apply to_array
x_test = to_array()
```

### RMSProp Model
Predicting and writing away results of a Convolution Neural Network with a RMSProp optimizer. 

```python
# Load model named 'RMSProp.h5'
rmsprop_test = keras.models.load_model('RMSProp.h5') 
```

```python
# For performance reasons the test set has to be split in five parts before predictions can be made
test = np.array([0, 11492, 22984, 34476, 45968, 57458])

results = []

# For loop to make predictions in five batches 
for i in range(len(test)-1):
    print(test[i], test[i+1])

    results.append(rmsprop_test.predict(x_test[test[i]:test[i+1]]))
```

```python
# Function to vertically merge the results and apply a label based on the assigned probability
def results_df (results = results):
    df_0 = pd.DataFrame(results[0])

    df_1 = pd.DataFrame(results[1])

    df_2 = pd.DataFrame(results[2])

    df_3 = pd.DataFrame(results[3])

    df_4 = pd.DataFrame(results[4])

    frames = [df_0, df_1, df_2, df_3, df_4]

    result_df = pd.concat(frames)

    result_df['Label'] = [1 if score >= 0.50 else 0 for score in result_df[1]] # Set the decision threshold at 0.5

    result_df.reset_index(inplace = True, drop=True)

    return result_df
```

```python
# Apply function results_df
results_rmsprop = results_df(results)
results_rmsprop
```

```python
# Write results to an external location
results_rmsprop.to_csv('rmsprop.csv', index=False)
```
